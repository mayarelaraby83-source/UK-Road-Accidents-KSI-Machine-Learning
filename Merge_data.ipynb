```python
import numpy as np
import pandas as pd
```

```python
acc = pd.read_csv('Accidents_2018.csv')
veh = pd.read_csv('Vehicles_2018.csv', low_memory=False)
```

```python
acc['Accident_Index'] = acc['Accident_Index'].astype(str)
veh['Accident_Index'] = veh['Accident_Index'].astype(str)
```

```python
veh['Age_of_Driver'] = veh['Age_of_Driver'].replace(-1, pd.NA)
veh['Engine_Capacity_(CC)'] = veh['Engine_Capacity_(CC)'].replace(-1, pd.NA)
veh['Age_of_Vehicle'] = veh['Age_of_Vehicle'].replace(-1, pd.NA)
```

```python
veh_agg = veh.groupby('Accident_Index').agg(
    veh_count=('Vehicle_Reference', 'count'),
    avg_driver_age=('Age_of_Driver', 'mean'),
    max_engine_cc=('Engine_Capacity_(CC)', 'max'),
    avg_vehicle_age=('Age_of_Vehicle', 'mean'),
    n_male_drivers=('Sex_of_Driver', lambda x: (x == 1).sum())
).reset_index()
```

```python
df = acc.merge(veh_agg, on='Accident_Index', how='left', validate='one_to_one')

df.to_csv('merged_accidents_vehicles.csv', index=False)
print(df.shape)
```
